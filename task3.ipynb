```python
 # Task 3: Exploratory Data Analysis (EDA) on Iris Dataset
 # Tools: Python, Matplotlib, Seaborn

 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
import warnings
warnings.filterwarnings('ignore')
```

```python
# STEP 0: Load Dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
 
print("=" * 55)
print("         IRIS DATASET - EDA REPORT")
print("=" * 55)
```

```python
# STEP 1: Summary Statistics
print("\n[1] DATASET SHAPE:", df.shape)
print("\n[2] FIRST 5 ROWS:")
print(df.head())
 
print("\n[3] DATA TYPES:")
print(df.dtypes)
 
print("\n[4] MISSING VALUES:")
print(df.isnull().sum())
 
print("\n[5] SUMMARY STATISTICS:")
print(df.describe().round(2))
 
print("\n[6] SPECIES COUNT:")
print(df['species'].value_counts())
 
print("\n[7] MEAN VALUES PER SPECIES:")
print(df.groupby('species').mean().round(2))
```

```python
# STEP 2: Visualize Distributions
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Feature Distributions by Species", fontsize=16, fontweight='bold')
 
features = iris.feature_names
colors = ['#4C72B0', '#DD8452', '#55A868']
 
for i, (ax, feature) in enumerate(zip(axes.flatten(), features)):
    for j, species in enumerate(iris.target_names):
        subset = df[df['species'] == species][feature]
        ax.hist(subset, bins=15, alpha=0.6, label=species, color=colors[j])
    ax.set_title(feature, fontweight='bold')
    ax.set_xlabel("Value (cm)")
    ax.set_ylabel("Frequency")
    ax.legend()
 
plt.tight_layout()
plt.savefig("plot1_distributions.png", dpi=150, bbox_inches='tight')
plt.close()
print("\n✅ Saved: plot1_distributions.png")
 
```

```python
# STEP 3: Correlation Heatmap
plt.figure(figsize=(8, 6))
corr = df.drop('species', axis=1).corr()
mask = corr.where(~(corr == 1)).isna()  # optional: keep diagonal
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            linewidths=0.5, square=True, cbar_kws={"shrink": 0.8})
plt.title("Correlation Heatmap - Iris Features", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig("plot2_heatmap.png", dpi=150, bbox_inches='tight')
plt.close()
print("✅ Saved: plot2_heatmap.png")
 
```

```python
# STEP 4: Detect Patterns - Pairplot

pairplot = sns.pairplot(df, hue='species', palette='Set2',
                        diag_kind='kde', plot_kws={'alpha': 0.7})
pairplot.fig.suptitle("Pairplot - Pattern Detection", y=1.02,
                      fontsize=14, fontweight='bold')
plt.savefig("plot3_pairplot.png", dpi=150, bbox_inches='tight')
plt.close()
print("✅ Saved: plot3_pairplot.png")
```

```python
# STEP 5: Identify Trends - Boxplots

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Boxplots - Trends per Species", fontsize=16, fontweight='bold')
 
for i, (ax, feature) in enumerate(zip(axes.flatten(), features)):
    sns.boxplot(data=df, x='species', y=feature,
                palette='Set2', ax=ax, width=0.5)
    ax.set_title(feature, fontweight='bold')
    ax.set_xlabel("")
    ax.set_ylabel("Value (cm)")
 
plt.tight_layout()
plt.savefig("plot4_boxplots.png", dpi=150, bbox_inches='tight')
plt.close()
print("✅ Saved: plot4_boxplots.png")
 
# ─────────────────────────────────────────
# BONUS: Violin Plot
# ─────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle("Violin Plots - Feature Spread", fontsize=14, fontweight='bold')
 
sns.violinplot(data=df, x='species', y='petal length (cm)',
               palette='muted', ax=axes[0])
axes[0].set_title("Petal Length Distribution")
 
sns.violinplot(data=df, x='species', y='sepal length (cm)',
               palette='muted', ax=axes[1])
axes[1].set_title("Sepal Length Distribution")
 
plt.tight_layout()
plt.savefig("plot5_violin.png", dpi=150, bbox_inches='tight')
plt.close()
print("✅ Saved: plot5_violin.png")
 
print("\n" + "=" * 55)
print("           KEY FINDINGS")
print("=" * 55)
print("""
1. DISTRIBUTIONS:
   - Petal length & width separate species very clearly
   - Setosa has distinctly smaller petal measurements
 
2. CORRELATIONS:
   - Petal length & petal width: very high correlation (~0.96)
   - Sepal width has low/negative correlation with others
 
3. PATTERNS (Pairplot):
   - Setosa is clearly separable from others
   - Versicolor and Virginica overlap slightly
 
4. TRENDS (Boxplots):
   - Virginica has the largest features overall
   - Setosa shows least variability in petal features
 
5. CONCLUSION:
   - Petal measurements are the best features for
     classifying Iris species
""")
 

```
